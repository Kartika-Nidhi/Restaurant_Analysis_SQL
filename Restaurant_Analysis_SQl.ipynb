# Global Restaurant Analytics & Relational Database Project

<div style = "background-color:lightblue; color:green; font-size:20px; text-align:justify">In the modern data ecosystem, raw data often arrives in unstructured, flat files packed with redundancy. This project simulates a real-world data engineering and analytics workflow by taking a flat CSV file containing 9,551 restaurant records and 21 attributes and transforming it into a clean, optimized relational database.

By applying database normalization principles (up to Third Normal Form), the flat dataset was decomposed into 4 to 5 interrelated tables (e.g., entities covering Restaurants, Locations, Cuisines, and Pricing/Services) connected via primary and foreign keys. Following schema design, advanced SQL queries—utilizing Common Table Expressions (CTEs), window functions, and multi-table joins—were executed to extract actionable business insights into global dining behaviors, cuisine popularity, pricing tiers, and the operational factors that drive customer ratings.
